**(a)**

```python
import numpy as np
# values
x1 = np.array([5, 15, 15, 25])
x2 = np.array([10,30,50,30])
x3 = np.array([5,15,15,5])

# mean = x / len(x)
x1mean = x1.sum() / len(x1)
x2mean = x2.sum() / len(x2)
x3mean = x3.sum() / len(x3)
print(f'x1mean= {x1mean}')
print(f'x2mean= {x2mean}')
print(f'x3mean= {x3mean}')

#var = (( x - xmean)**2) / len(x)
x1var = ((x1 - x1mean)**2).sum() / len(x1)
x2var = ((x2 - x2mean)**2).sum() / len(x2)
x3var = ((x3 - x3mean)**2).sum() / len(x3)
print(f'x1var= {x1var}')
print(f'x2var= {x2var}')
print(f'x3var= {x3var}')

```

**(b)**

```python
total_var = x1var + x2var + x3var
print(f'total_var= {total_var}')
```
